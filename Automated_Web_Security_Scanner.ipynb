## 2. Scanner code
Change `url` to a website you own or have permission to test. This version intentionally avoids `input()` so it runs smoothly in Colab on mobile.

```python
import requests
from urllib.parse import urlparse

SECURITY_HEADERS = {
    "Content-Security-Policy": "Helps control which resources a page can load.",
    "Strict-Transport-Security": "Encourages browsers to use secure HTTPS connections.",
    "X-Content-Type-Options": "Helps prevent MIME-type confusion.",
    "X-Frame-Options": "Can help protect against clickjacking.",
    "Referrer-Policy": "Controls referrer information sent by the browser.",
}

def scan_website(url):
    parsed = urlparse(url)
    if parsed.scheme not in ("http", "https") or not parsed.hostname:
        print("Please set url to a complete http:// or https:// address.")
        return

    print("=" * 48)
    print("AUTOMATED WEB SECURITY SCANNER")
    print("=" * 48)
    print("Website:", url)

    try:
        response = requests.get(
            url,
            timeout=10,
            allow_redirects=True,
            headers={"User-Agent": "Beginner-Web-Security-Scanner/1.0"},
        )
    except requests.RequestException as error:
        print("Connection failed:", error)
        return

    headers = {name.lower(): value for name, value in response.headers.items()}
    present = []
    missing = []

    print("HTTP status:", response.status_code)
    print("Final URL:", response.url)
    print("\n--- SECURITY HEADER RESULTS ---")

    for name, purpose in SECURITY_HEADERS.items():
        if name.lower() in headers:
            present.append(name)
            print("[PRESENT]", name)
        else:
            missing.append(name)
            print("[MISSING]", name)
            print("Purpose:", purpose)

    print("\n--- SCAN SUMMARY ---")
    print("Headers checked:", len(SECURITY_HEADERS))
    print("Present:", len(present))
    print("Missing:", len(missing))
    print("Note: Missing headers need context and manual review.")
    print("This is a basic header checker, not a full vulnerability scanner.")
    print("Scan completed.")

# Demonstration target; change only to a permitted website.
url = "https://example.com"
scan_website(url)
```

## 3. How to interpret results
- HTTP status `200` means the server successfully responded to the request.
- `PRESENT` means the selected header appeared in the response.
- `MISSING` means it was not found in that response; this alone does not prove the website is vulnerable.
- This project checks response headers only. It does not perform exploitation or a full security assessment.
